```python
#categorical and numerical columns
categorical_cols = df.select_dtypes(include=['object']).columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Categorical Columns:", categorical_cols)
print("Numerical Columns:", numerical_cols)
```

Data Analysis

```python
## Visualizations fot numerical columns
#boxplot
for num_col in numerical_cols:
    plt.figure(figsize=(10, 4))
    sns.boxplot(df[num_col])
    plt.title(f'Boxplot of {num_col}')
    plt.xlabel(num_col)
    plt.ylabel("count")
    plt.show()
```

```python
## Visualizations fot numerical columns

for num_col in numerical_cols:
    plt.figure(figsize=(10, 4))
    sns.histplot(df[num_col], kde=True, bins=30)
    plt.title(f'Distribution of {num_col}')
    plt.xlabel(num_col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
for cat_col in categorical_cols:
    plt.figure(figsize=(12, 6))
    sns.countplot(y=df[cat_col], order=df[cat_col].value_counts().index)
    plt.title(f'Countplot of {cat_col}')
    plt.xlabel('Count')
    plt.ylabel(cat_col)
    plt.show()
```

```python
#Label Encoding
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

mapping_dict = {}

for col in categorical_cols:
    df[col] = le.fit_transform(df[col])
    mapping_dict[col] = dict(zip(le.classes_, le.transform(le.classes_)))

```

```python
##multicollinearity check

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

X = add_constant(df[numerical_cols])

vif_data = pd.DataFrame()
vif_data["Feature"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
vif_data = vif_data.sort_values(by="VIF", ascending=False)
print(vif_data)
```

no muticollinearity

```python
#heatmap for correlation
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
##skewness and kurtosis
from scipy.stats import skew, kurtosis
skewness = df[numerical_cols].apply(skew).sort_values(ascending=False)
kurtosis_values = df[numerical_cols].apply(kurtosis).sort_values(ascending=False)
print("Skewness:\n", skewness)
print("\nKurtosis:\n", kurtosis_values)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = df.select_dtypes(include=['int64', 'float64']).columns
stats_df = pd.DataFrame({
    'Skewness': df[num_cols].skew(),
    'Kurtosis': df[num_cols].kurt()
}).reset_index().rename(columns={'index': 'Feature'})

plt.figure(figsize=(12,6))
sns.barplot(x='Feature', y='Skewness', data=stats_df)
plt.xticks(rotation=90)
plt.title("Skewness of Numeric Features")
plt.show()

plt.figure(figsize=(12,6))
sns.barplot(x='Feature', y='Kurtosis', data=stats_df)
plt.xticks(rotation=90)
plt.title("Kurtosis of Numeric Features")
plt.show()
```
